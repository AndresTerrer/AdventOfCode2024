````python
# Approach:
""" 
For every 0, get a list of neighbours with elevation 0 + 1 recursively, then
when the elevation is 9, return the length of the recursive list's last step, that is
the score of this 0.

Repeat for each 0
"""
````

````python
import numpy as np
class Day10:
    def __init__(self, txt_path: str):
        self.topo_map = read_as_character_array(txt_path).astype(int)
        self.hiking_map = np.ones_like(self.topo_map) * -1

        pass
    def __str__(self):
        """ 
        Print map
        """
        return str(
            print(
                "Topographic map: \n",self.topo_map.view(),
                "\n\n Hikes Map: \n", self.hiking_map.view()
            )
        )
    
    def find_hikes(self):
        """ 
        Find all hinkes and write them into the hiking map for debugging.
        """

        pass
    
    def get_next_hinking_steps(self, position: tuple) -> list[tuple]:
        """ 
        Given a location on the map, retrieve the lists of possible next steps,
        where elevation goes up by one
        """
        start_elevation = self.get_elevation(position)
        if start_elevation == 9:
            return []
        
        neighbours = self.get_neighbours(position)
        n_elevations = [self.get_elevation(n) for n in neighbours]
    
        return [
            n for n, e in zip(neighbours, n_elevations) 
            if (e == start_elevation + 1)
        ]

    def get_elevation(self, position: tuple) -> int:
        """ get the elevation on the map"""
        try:
            return self.topo_map[*position]
        except Exception as e:
            print(e, "Position: ", position)
    
    def get_neighbours(self, position) -> list[tuple]:
        """ return valid neighbour indeces for a postion"""

        n, m = self.topo_map.shape

        # All posible indeces for neighbours:
        directions = [
            (1, 0), (-1, 0),
            (0, 1), (0, -1),
        ]
        neighbours = [
            tuple(map(lambda i, j: i + j, position, d)) 
            for d in directions
        ]

        # for k, neighbour in enumerate(neighbours):
        #     i, j = neighbour[0], neighbour[1]

        #     if i < 0 or i >= n or j < 0 or j >= m:
        #         print("pop neighbour: ", k, neighbour)

        # NOTE: This changes the lenght of the list as it iterates, so its a no-go
        #         neighbours.pop(k)

        # Update the whole list instead, with possitive logic.
        neighbours = [(i, j) for i, j in neighbours if 0 <= i < n and 0 <= j < m]

        return neighbours
    
    def update_hinking_map(self, position: tuple):
        """ Write a position on the hinking map """
        self.hiking_map[*position] = self.topo_map[*position]

day10 = Day10(test_path)
print(day10)
day10.get_next_hinking_steps((0,2))
````
